# Loan Details — Silver Layer

**Task 3 of the `banking_data_workflow` Lakeflow job.**

This notebook builds the `loan_details` table (product, account, and loan attributes) from `bank_master_data_bronze`, complementing `borrower_details` (customer demographics).

* Upstream: `bank_master_data_bronze` (from `01_ingest_banking`)
* Output table: `loan_details` (~5,000 rows)

```sql
-- Task 3 of banking_data_workflow: build loan attributes from the bronze table
CREATE OR REPLACE TABLE loan_details AS
SELECT
    ID,
    Mortgage,
    CCAvg,
    Personal_Loan,
    Securities_Account,
    CD_Account,
    Online,
    CreditCard
FROM bank_master_data_bronze;

select count(*) from loan_details
```
