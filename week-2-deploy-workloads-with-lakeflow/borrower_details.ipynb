# Borrower Details — Silver Layer

**Task 2 of the `banking_data_workflow` Lakeflow job.**

This notebook builds the `borrower_details` table (customer demographics: age, income, education, family) from `bank_master_data_bronze`, complements `loan_details` (loan/account attributes). It also runs two quality checks:

* Count of eligible borrowers using the `min_income` parameter (job can pass it as a task parameter; default 100)
* Validation of the upstream row count (`bronze_count`) published as a task value by `01_ingest_banking`

* Upstream: `bank_master_data_bronze` (from `01_ingest_banking`)
* Output table: `borrower_details` (~5,000 rows)

```sql
-- Task 2 of banking_data_workflow: build borrower demographics from the bronze table
CREATE OR REPLACE TABLE borrower_details AS
SELECT
    ID,
    Age,
    Experience,
    Income as Income,
    ZIP_Code,
    Family,
    Education
FROM bank_master_data_bronze;

select count(*) from borrower_details
```

```sql
-- Parameterized quality check: count eligible borrowers by min_income (job parameter)
CREATE WIDGET TEXT min_income DEFAULT "100";

SELECT COUNT(*) AS eligible_borrowers
FROM borrower_details
WHERE Income >= :min_income;
```

```sql
-- Validate the upstream row count passed from 01_ingest_banking (:bronze_count)
SELECT CAST(:bronze_count AS BIGINT) AS upstream_count;
```
